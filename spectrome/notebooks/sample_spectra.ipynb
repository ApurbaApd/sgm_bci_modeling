```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import welch

col_rs = "#16ACF2" 
col_mi = "#FA50C4"

fs = 1000  
t = np.arange(0, 2, 1/fs)

# First signal 
signal1 = np.sin(2 * np.pi * 10 * t) + 0.5 * np.random.randn(len(t))
f1, Pxx1 = welch(signal1, fs, nperseg=1024)

# Second signal 
signal2 = np.sin(2 * np.pi * 20 * t) + 0.5 * np.random.randn(len(t))
f2, Pxx2 = welch(signal2, fs, nperseg=1024)

fig, axs = plt.subplots(2, 1, figsize=(6, 4), sharex=True, dpi=400)

axs[0].semilogy(f1, Pxx1, color=col_rs)
axs[0].axis('off')


axs[1].semilogy(f2, Pxx2, color=col_mi)
axs[1].axis('off')

plt.subplots_adjust(wspace=0, hspace=0)
plt.margins(0, 0)
plt.tight_layout(pad=0)
# plt.savefig(f"/Users/apurbadebnath/Desktop/sgm_bci_paper_figures/spectra_sample.png", dpi=400)
plt.show()
```
